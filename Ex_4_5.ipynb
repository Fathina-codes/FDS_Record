```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import gutenberg

# Download necessary NLTK data
nltk.download('gutenberg')
nltk.download('punkt')
nltk.download('punkt_tab')  # <-- ADD THIS LINE

# Load raw text from Gutenberg corpus
sample = gutenberg.raw("austen-emma.txt")

# Tokenize the sample text
token = word_tokenize(sample)

# Store the first 50 tokens
wlist = []
for i in range(50):
    wlist.append(token[i])

# Calculate frequency of each word in the list
wordfreq = [wlist.count(w) for w in wlist]

# Print the word-frequency pairs
print("Pairs\n" + str(list(zip(wlist, wordfreq))))
```
