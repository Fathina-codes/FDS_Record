```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler, MinMaxScaler

# Direct input creation (pre-process_datasample.csv)
data = {
    'Country': ['France', 'Spain', 'Germany', 'Spain', 'Germany', 'France', 'Spain', 'France', np.nan, 'France'],
    'Age': [44.0, 27.0, 30.0, 38.0, 40.0, 35.0, np.nan, 48.0, 50.0, 37.0],
    'Salary': [72000.0, 48000.0, 54000.0, 61000.0, np.nan, 58000.0, 52000.0, 79000.0, 83000.0, 67000.0],
    'Purchased': ['No', 'Yes', 'No', 'No', 'Yes', 'Yes', 'No', 'Yes', 'No', 'Yes']
}

df = pd.DataFrame(data)

print("--- Initial DataFrame ---")
print(df)

print("\n--- df.head() ---")
print(df.head())

# Fill missing categorical values for Country
df.Country.fillna(df.Country.mode()[0], inplace=True)

# Separate features and label
features = df.iloc[:, :-1].values
label = df.iloc[:, -1].values

# Impute missing numerical values (Age & Salary)
age_imputer = SimpleImputer(strategy="mean", missing_values=np.nan)
salary_imputer = SimpleImputer(strategy="mean", missing_values=np.nan)

features[:, [1]] = age_imputer.fit_transform(features[:, [1]])
features[:, [2]] = salary_imputer.fit_transform(features[:, [2]])

print("\n--- Features after Imputation ---")
print(features)

# OneHotEncode the 'Country' column
oh = OneHotEncoder(sparse_output=False)
Country = oh.fit_transform(features[:, [0]])

print("\n--- OneHotEncoded Country ---")
print(Country)

# Combine OneHotEncoded Country and numerical features (Age, Salary)
final_set = np.concatenate((Country, features[:, [1, 2]]), axis=1)

print("\n--- Final Set Before Scaling ---")
print(final_set)

# StandardScaler
sc = StandardScaler()
feat_standard_scaler = sc.fit_transform(final_set)

print("\n--- Standard Scaler Output ---")
print(feat_standard_scaler)

# MinMaxScaler
mms = MinMaxScaler(feature_range=(0, 1))
feat_minmax_scaler = mms.fit_transform(final_set)

print("\n--- MinMax Scaler Output ---")
print(feat_minmax_scaler)
```
