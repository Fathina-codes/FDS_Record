```python
import numpy as np
import pandas as pd
data = {
    'emp_id': [101, 102, 103, 104, 105],
    'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'salary': [75000, 62000, 85000, 54000, 91000]
}
df= pd.DataFrame(data)
```

```python
print(df.head())
print(df.tail())
print(df.info())
print(df.salary)
print(type(df.salary))
print("Mean:", df.salary.mean())
print("Median:", df.salary.median())
print("Mode:\n", df.salary.mode())
print("Variance:", df.salary.var())
print("Std Dev:", df.salary.std())
print(df.describe())
print(df.describe(include="all"))
```

```python
empCol=df.columns
print(empCol)
emparray=df.values
print(emparray)
```
