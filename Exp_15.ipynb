```python
import numpy as np
import pandas as pd

# Direct input creation
data = {
    'Country': ['France', 'Spain', 'Germany', 'Spain', 'Germany', 'France', 'Spain', 'France', np.nan, 'France'],
    'Age': [44.0, 27.0, 30.0, 38.0, 40.0, 35.0, np.nan, 48.0, 50.0, 37.0],
    'Salary': [72000.0, 48000.0, 54000.0, 61000.0, np.nan, 58000.0, 52000.0, 79000.0, 83000.0, 67000.0],
    'Purchased': ['No', 'Yes', 'No', 'No', 'Yes', 'Yes', 'No', 'Yes', 'No', 'Yes']
}

df = pd.DataFrame(data)

print("--- Initial DataFrame ---")
print(df)

print("\n--- Initial df.info() ---")
df.info()

# Mode, Median, and Mean Imputation (clean assignments without inplace)
df['Country'] = df['Country'].fillna(df['Country'].mode()[0])
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Salary'] = df['Salary'].fillna(round(df['Salary'].mean()))

print("\n--- DataFrame After Imputation ---")
print(df)

# One-hot encoding Country and concatenating
updated_dataset = pd.concat([pd.get_dummies(df['Country']), df.iloc[:, [1, 2, 3]]], axis=1)

print("\n--- pd.get_dummies(df.Country) ---")
print(pd.get_dummies(df['Country']))

print("\n--- Updated Dataset (Before Replacing Purchased) ---")
print(updated_dataset)

print("\n--- Cleaned df.info() ---")
df.info()

# Label Encoding Purchased column ('No': 0, 'Yes': 1)
updated_dataset['Purchased'] = updated_dataset['Purchased'].replace({'No': 0, 'Yes': 1}).astype(int)

print("\n--- Final Updated Dataset ---")
print(updated_dataset)
```
