```python
import pandas as pd
import matplotlib.pyplot as plt
import io

# Direct CSV data inside the code
csv_data = """Month,Sales
Jan,100
Feb,150
Mar,120
Apr,180"""

# Read CSV directly from string
df = pd.read_csv(io.StringIO(csv_data))

# Plot line graph
plt.plot(df['Month'], df['Sales'], marker='o')

# Labels and title
plt.xlabel('Month')
plt.ylabel('Sales')
plt.title('Line Plot')

# Show plot
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Direct data in pandas DataFrame
df = pd.DataFrame({'Subject': ['Math', 'Science', 'English', 'History'], 'Marks': [85, 92, 78, 88]})

# Plot bar chart
plt.bar(df['Subject'], df['Marks'])

# Labels and title
plt.xlabel('Subject')
plt.ylabel('Marks')
plt.title('Student Performance')

# Show plot
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Direct data in pandas DataFrame
df = pd.DataFrame({'Fruit': ['Apples', 'Bananas', 'Cherries', 'Grapes'], 'Quantity': [35, 25, 15, 25]})

# Plot pie chart
plt.pie(df['Quantity'], labels=df['Fruit'], autopct='%1.1f%%')

# Title
plt.title('Fruit Distribution')

# Show plot
plt.show()
```
